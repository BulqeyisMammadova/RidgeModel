```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression,Ridge
```

```python
df = pd.read_csv('student_scores.csv')
df.head(3)
```

```python
y = df[['Scores']]
X = df[['Hours']]
```

```python


plt.style.use('fivethirtyeight')
plt.figure(figsize=(10,5))
plt.scatter(X,y)
plt.xlabel('Hours')
plt.ylabel('Scores')
plt.show()
```

```python
lm = LinearRegression()
model = lm.fit(X,y)
```

```python
model.score(X,y)
```

```python
alphas = [1,10,100,200]
for a in alphas:
    r = Ridge(alpha=a)
    model_r = r.fit(X,y)
    print('Skore: ',model_r.score(X,y))
    print('Coef: ',model_r.coef_)
```
